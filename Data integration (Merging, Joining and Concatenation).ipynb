```python
import pandas as pd
```

## CHALLENGE 1: Creating 3 DataFrames

### DataFrame 1: Student Info

#### Creating data dictionary for the DataFrame

```python
student_info = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104], 
    "Name": ["Laura", "Benedine", "Ato", "Newman"],
    "Gender": ["F", "M", "M", "M"]
})
```

```python
pd.DataFrame(student_info)
```

### DataFrame 2: Test Scores

#### Creating data dictionary for the DataFrame

```python
test_scores = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104], 
    "Math_Score": [88, 92, 79, 85],
    "English_Score": [90, 87, 75, 89]
})
```

```python
pd.DataFrame(test_scores)
```

### DataFrame 3: Attendance

#### Creating data dictionary for the DataFrame

```python
attendance_data = pd.DataFrame({
    "Student_ID": [101, 102, 103, 104], 
    "Days_Present": [22, 18, 20, 25],
    "Total_Days": [25, 25, 25, 25]
})
```

```python
pd.DataFrame(attendance_data)
```

## CHALLENGE 2: Concatenate Vertically

### DataFrame 4: more_students

```python
more_students = pd.DataFrame({
    "Student_ID": [105, 106], 
    "Name": ["Lauren", "Loice"],
    "Gender": ["F", "F"]
})
```

```python
pd.DataFrame(more_students)
```

### Concatenating 2 DataFrames(students_info and more_students)

```python
pd.concat([student_info, more_students], ignore_index=True)
all_students = pd.concat([student_info, more_students], ignore_index=True)
```

```python
all_students
```

### CHALLENGE 3: 2 DataFrames
Merge students_info with test_scores

```python
pd.merge(student_info, test_scores, on="Student_ID", how="outer")
merged_scores = pd.merge(student_info, test_scores, on="Student_ID", how="outer")
```

```python
merged_scores
```

```python
pd.merge(student_info, test_scores, on="Student_ID", how="inner")
merged_scores = pd.merge(student_info, test_scores, on="Student_ID", how="inner")
```

```python
merged_scores
```

### CHALLENGE 4: Merging new DataFrame with Attendance

```python
final_data = pd.merge(merged_scores, attendance_data, on="Student_ID", how="inner")
```

```python
final_data
```

### CHALLENGE 5: Cleaning the Data
Filling missing values for clarity

```python
Cleaned_data = final_data.fillna("Missing")
```

```python
Cleaned_data
```
